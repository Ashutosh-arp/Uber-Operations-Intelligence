### 2.1.1 Calculate Business KPIs

```python
total_bookings = len(df)

completed = (df["Booking Status"] == "Completed").sum()
cancelled_customer = (df["Booking Status"] == "Cancelled by Customer").sum()
cancelled_driver = (df["Booking Status"] == "Cancelled by Driver").sum()
no_driver = (df["Booking Status"] == "No Driver Found").sum()
incomplete = (df["Booking Status"] == "Incomplete").sum()

completion_rate = completed / total_bookings * 100
customer_cancelled_rate = cancelled_customer / total_bookings * 100
driver_cancelled_rate = cancelled_driver / total_bookings * 100
no_driver_rate = no_driver / total_bookings * 100
incompleted_rate = incomplete / total_bookings * 100

total_revenue = df["Booking Value"].sum()
average_booking_values = ( df["Booking Value"].mean())
```

### KPI Summary

```python
print("=" * 60)
print("UBER BUSINESS OVERVIEW")
print("=" * 60)

print(f"Total Bookings          : {total_bookings:,}")
print(f"Completed Rides         : {completed:,}")
print(f"Completion Rates        : {completion_rate:.2f}%")
print(f"Customer Cancellation   : {customer_cancelled_rate:.2f}%")
print(f"Driver Cancellation     : {driver_cancelled_rate:.2f}%")
print(f"No Driver Found         : {no_driver_rate:.2f}%")
print(f"Incomplete Ride Rates   : {incompleted_rate:.2f}%")
print(f"Total Revenue           : Rs {total_revenue:,.2f}")
print(f"Average Booking Value   : Rs {average_booking_values:.2f}")
```

## 2.4.1 Prepare Time-Based Features

### Objective

To Create the time features required for the analysis.

```python
df[["Date","Time"]].dtypes
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Time"] = pd.to_datetime(df["Time"], format= "%H:%M:%S")
```

```python
df["Hour"] = df["Time"].dt.hour
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Month_Name"] = df["Date"].dt.month_name()
df["Day_Name"] = df["Date"].dt.day_name()
```

```python
df["Day_Type"] = np.where(df["Date"].dt.dayofweek <5,"Weekday","Weekend")
```

```python
df[[ 
     "Date",
     "Time",
     "Hour",
     "Day_Name",
     "Month_Name",
     "Day_Type"
]].head()
```

## 2.5.3 Location Concentration

```python
total_bookings = len(df)

top_10_pickup_share = (pickup_counts.head(10).sum()/total_bookings)*100
top_10_drop_share = (drop_counts.head(10).sum()/total_bookings)*100

print(f"Top 10 Pickup Location Share : {top_10_pickup_share:.2f}%")
print(f"Top 10 Drop Location Share : {top_10_drop_share:.2f}%")
```

## 2.5.4 Pickup vs Drop Location Comparison

### Objective

Compare pickup and drop activity across the most frequently used locations.

```python
top_locations = (set(pickup_counts.head(10).index).union(set(drop_counts.head(10).index)))

location_comparison = pd.DataFrame({"Pickup Bookings": pickup_counts,"Drop Bookings": drop_counts}).fillna(0)

location_comparison = (location_comparison.loc[list(top_locations)].sort_values("Pickup Bookings", ascending=False).head(10))

location_comparison
```

```python
ax = location_comparison.plot(kind="bar",figsize=(12,6))

plt.title("Pickup vs Drop Booking Activity")
plt.xlabel("Location")
plt.ylabel("Number of Bookings")

plt.xticks(rotation=30, ha="right")

plt.grid(axis="y",alpha=0.3)

plt.legend(title = "Booking Type",loc="upper left",bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

## Business Insight

- **Demand is widely distributed** across locations rather than being concentrated in a small number of locations.
- The top 10 pickup locations account for approximately **6.16%** of total bookings, while the top 10 drop account for approximately **6.08%**,         indicating relatively low geographic concentration.
- The pickup vs Drop comparison also shows that the highest demand locations generally have **slightly more pickup bookings that drop bookings**,           suggesting that these locations generates consistently high trip activity on both ends of the journey.
- Overall, the business appears to have a **broadly distributed demand network**, meaning operational planning should comnsider multiple locations rather   than focusing resources on only a few high-volume locations.

## 2.6.2 Cancellation Rate by Vehicle Type

Identify vehicle types with the highest cancellation rates.

```python
vehicle_total=(df.groupby("Vehicle Type").size())

vehicle_cancelled = (df[df["Booking Status"].isin(["Cancelled by Customer","Cancelled by Driver"])].groupby("Vehicle Type").size())

vehicle_cancellation_rate = (vehicle_cancelled.div(vehicle_total).mul(100).sort_values(ascending=False))

vehicle_cancellation_rate
```

This suggests that cancellations are not concentrated in one specific time period or vehicle category, but represents a broader       operational issue.

- Overall, the business should prioritize **reducing driver-side cancellations and improving ride fulfillment**, as these issues directly reduces       successfull completed rides and represents a larger operational opportunity than customer cancellations alone.
